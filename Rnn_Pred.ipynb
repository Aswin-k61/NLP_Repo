```python
import numpy as np

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding
from tensorflow.keras.layers import SimpleRNN
from tensorflow.keras.layers import Dense


```

```python
data = """
machine learning is fun
machine learning is powerful
machine learning is widely used
machine learning improves predictions
machine learning helps automation

deep learning uses neural networks
deep learning is a subset of machine learning
deep learning works with large data
deep learning models are complex
deep learning improves accuracy

python is used for machine learning
python is popular for data science
python supports many libraries
python is easy to learn
python is powerful for AI

artificial intelligence is transforming technology
artificial intelligence is the future
artificial intelligence automates tasks
artificial intelligence improves systems
artificial intelligence uses machine learning

neural networks mimic human brain
neural networks learn patterns
neural networks require data
neural networks improve predictions
neural networks are used in deep learning

data science uses statistics and machine learning
data science extracts insights
data science works with big data
data science helps decision making
data science uses python

AI is used in healthcare
AI is used in finance
AI is used in robotics
AI is used in automation
AI is used in education

big data is important for machine learning
big data improves models
big data requires storage
big data is growing rapidly
big data is used in AI

natural language processing uses machine learning
natural language processing understands text
natural language processing is part of AI
natural language processing helps chatbots
natural language processing analyzes language

computer vision uses deep learning
computer vision processes images
computer vision is used in self driving cars
computer vision detects objects
computer vision improves recognition
"""
```

```python
tokenizer=Tokenizer()

tokenizer.fit_on_texts(
[data]
)

total_words=len(
tokenizer.word_index)+1
print(total_words)
```

```python
input_sequences=[]
for line in data.split("\n"):
    token_list=tokenizer.texts_to_sequences(
      [line]
    )[0]
    for i in range(1,len(token_list)):
      seq=token_list[:i+1]
      input_sequences.append(seq)
```

```python
max_len=max(
  [len(x) for x in input_sequences]
)

input_sequences=np.array(
    pad_sequences(
        input_sequences,
        maxlen=max_len,
        padding="pre"
    )
)
```

```python
X=input_sequences[:,:-1]
y=input_sequences[:,-1]
```

```python
model=Sequential()
model.add(Embedding(
    total_words,
  10,
  input_length=max_len-1
))

model.add(
    SimpleRNN(32)
)
model.add(Dense(
    total_words,
    activation="softmax"
))
```

```python
model.compile(
    loss="sparse_categorical_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)
```

```python
model.fit(
    X,
    y,
    epochs=300,
    verbose=1
)
```

```python
send_text="machine learning is"

token_list = tokenizer.texts_to_sequences([send_text])[0]

token_list=pad_sequences(
    [token_list],
    maxlen=max_len-1,
    padding='pre')

pred=np.argmax(
    model.predict(
        token_list,verbose=0),
    axis=1
)[0]

for word,index in tokenizer.word_index.items():
    if index==pred:
        print("Next word",word)
```
