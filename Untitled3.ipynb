```python
import numpy as np
import pandas as pd
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding
from tensorflow.keras.layers import SimpleRNN
from tensorflow.keras.layers import Dense
```

```python
df=pd.read_csv('/content/Quotes Dataset.csv')
df.head()
```

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts(df['Quote'])
total_words = len(tokenizer.word_index) + 1
print(f"Total words: {total_words}")

input_sequences = []
for line in df['Quote']:
    token_list = tokenizer.texts_to_sequences([line])[0]
    for i in range(1, len(token_list)):
        seq = token_list[:i+1]
        input_sequences.append(seq)
```

```python
print(f"Number of input sequences: {len(input_sequences)}")
print("First 5 input sequences:")
for i in range(min(5, len(input_sequences))):
    print(input_sequences[i])
```

```python
max_len=max(
  [len(x) for x in input_sequences]
)

input_sequences=np.array(
    pad_sequences(
        input_sequences,
        maxlen=max_len,
        padding="pre"
    )
)
```

```python
X=input_sequences[:,:-1]
y=input_sequences[:,-1]
```

```python
model=Sequential()
model.add(Embedding(
    total_words,
  10,
  input_length=max_len-1
))

model.add(
    SimpleRNN(32)
)
model.add(Dense(
    total_words,
    activation="softmax"
))
```

```python
model.compile(
    loss="sparse_categorical_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)
```

```python
model.fit(
    X,
    y,
    epochs=60,
    verbose=1
)
```

```python
send_text="The truth will"
num_words_to_predict = 5 # You can change this to predict more or fewer words

for _ in range(num_words_to_predict):
    token_list = tokenizer.texts_to_sequences([send_text])[0]

    token_list=pad_sequences(
        [token_list],
        maxlen=max_len-1,
        padding='pre')

    pred=np.argmax(
        model.predict(
            token_list,verbose=0),
        axis=1
    )[0]

    output_word = ""
    for word,index in tokenizer.word_index.items():
        if index==pred:
            output_word = word
            break
    send_text += " " + output_word

print("Generated text:", send_text)
```
