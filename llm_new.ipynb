```python
from transformers import pipeline

generator=pipeline('text-generation',model='google/flan-t5-small',max_new_tokens=100)
print("Model loaded successfully")
```

```python
def ask_llm(question):
  response = generator(question)
  return response[0]["generated_text"]

question = "Explain machine learning in simple words."
answer = ask_llm(question)

print("Question:", question)
print("Answer:", answer)


```

```python
from transformers import pipeline

en_to_ml = pipeline(
    "translation_en_to_ml",
    model="Helsinki-NLP/opus-mt-en-ml"
)

ml_to_en = pipeline(
    "translation_ml_to_en",
    model="Helsinki-NLP/opus-mt-ml-en"
)
```

```python
summarizer = pipeline(
    "summarization",
    model="facebook/bart-large-cnn"
)
```

```python
text = "Artificial Intelligence is transforming industries by enabling machines to learn from data."

summary = summarizer(text, max_length=50, min_length=10, do_sample=False)
print(summary[0]['summary_text'])

translated = en_to_ml(summary[0]['summary_text'])
print(translated[0]['translation_text'])
```

```python
print("AI Chatbot (Summarize + Translate)")
print("Commands:")
print("1. summarize: <text>")
print("2. translate en to ml: <text>")
print("3. summarize ml: <text>")
print("Type 'exit' to quit\n")

while True:
    user_input = input("You: ")

    if user_input.lower() == "exit":
        print("Bot: Goodbye!")
        break

    # 🔹 Summarize English
    elif user_input.lower().startswith("summarize:"):
        text = user_input[len("summarize:"):]

        summary = summarizer(text, max_length=50, min_length=10, do_sample=False)
        print("Bot:", summary[0]['summary_text'])

    # 🔹 Translate English → Malayalam
    elif user_input.lower().startswith("translate en to ml:"):
        text = user_input[len("translate en to ml:"):]

        translated = en_to_ml(text)
        print("Bot:", translated[0]['translation_text'])

    # 🔹 Malayalam → Summarize → Malayalam
    elif user_input.lower().startswith("summarize ml:"):
        text = user_input[len("summarize ml:"):]

        # Step 1: ML → EN
        english = ml_to_en(text)[0]['translation_text']

        # Step 2: Summarize
        summary = summarizer(english, max_length=50, min_length=10, do_sample=False)[0]['summary_text']

        # Step 3: EN → ML
        final = en_to_ml(summary)[0]['translation_text']

        print("Bot:", final)

    else:
        print("Bot: Invalid command")
```
